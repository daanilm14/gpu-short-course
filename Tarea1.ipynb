# Tarea 1: Configuración y Gráficas
Este notebook cumple con la Task 1 del curso.
Funciones graficadas: $f(x) = x^2$ y $g(x) = 2^x$.

```python
import numpy as np
import matplotlib.pyplot as plt

# 1.5 Crear vector x con 20 elementos
x = np.linspace(-1, 1, 20)

# 1.6 Graficar f(x)=x^2 y g(x)=2^x
plt.plot(x, x**2, label='$f(x)=x^2$', marker='o')
plt.plot(x, 2**x, label='$g(x)=2^x$', marker='s')

plt.legend()
plt.grid(True)
plt.title("Gráfica Task 1")
plt.show()
```
