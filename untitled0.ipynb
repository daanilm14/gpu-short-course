```python
import numpy as np
import matplotlib.pyplot as plt
import pycuda.driver as cuda
import pycuda.autoinit
from pycuda.compiler import SourceModule

# Implementación de la Ecuación de Calor en 2D (CPU)
def heatEquation2D(u0, num_timesteps, kappa, dt, dx, dy):
    u1 = np.empty_like(u0)
    ny, nx = u0.shape

    for n in range(num_timesteps):
        # Iteramos únicamente sobre las celdas internas, dejando fijos los bordes
        for j in range(1, ny - 1):
            for i in range(1, nx - 1):
                d2u_dx2 = (u0[j, i-1] - 2*u0[j, i] + u0[j, i+1]) / (dx*dx)
                d2u_dy2 = (u0[j-1, i] - 2*u0[j, i] + u0[j+1, i]) / (dy*dy)
                u1[j, i] = u0[j, i] + kappa * dt * (d2u_dx2 + d2u_dy2)

        # Mantener las condiciones de frontera fijas (constantes)
        # Borde superior e inferior
        u1[0, :] = u0[0, :]
        u1[ny-1, :] = u0[ny-1, :]
        # Borde izquierdo y derecho
        u1[:, 0] = u0[:, 0]
        u1[:, nx-1] = u0[:, nx-1]

        # Intercambio de temperaturas
        u0, u1 = u1, u0

    return u0
```

```python
# Configuración de parámetros
nx = 128
ny = 128
dx = 0.1
dy = 0.1
kappa = 10.0

# El paso de tiempo dt debe cumplir con la restricción de estabilidad CFL en 2D:
# dt <= (dx^2 * dy^2) / (2 * kappa * (dx^2 + dy^2))
dt = 0.0002  # Valor seguro y estable para dx=dy=0.1 y kappa=10
num_timesteps = 800

# Inicialización de la temperatura (Bordes a 20 grados)
initial_temp = np.full((ny, nx), 20.0, dtype=np.float32)

# Crear un círculo caliente en el centro (Radio N/6, Temperatura 100 grados)
r2 = (nx / 6)**2
center_x, center_y = nx / 2, ny / 2
for j in range(ny):
    for i in range(nx):
        if (i - center_x)**2 + (j - center_y)**2 < r2:
            initial_temp[j, i] = 100.0

# Resolver usando CPU
u_cpu = heatEquation2D(initial_temp.copy(), num_timesteps, kappa, dt, dx, dy)

# Visualizar resultado CPU
fig, axes = plt.subplots(1, 2, figsize=(12, 5))
im0 = axes[0].imshow(initial_temp, cmap='jet', origin='lower')
axes[0].set_title("Temperatura Inicial (CPU)")
fig.colorbar(im0, ax=axes[0], label="°C")

im1 = axes[1].imshow(u_cpu, cmap='jet', origin='lower')
axes[1].set_title(f"Temperatura Actualizada (CPU - {num_timesteps} pasos)")
fig.colorbar(im1, ax=axes[1], label="°C")
plt.show()
```

```python
heat_eqn_2d_kernel_src = """
__global__ void heatEqn2D(
    float* u1,
    const float* u0,
    float kappa,
    float dx,
    float dy,
    float dt,
    unsigned int nx,
    unsigned int ny) {

    // Coordenadas del hilo en 2D
    int i = blockIdx.x * blockDim.x + threadIdx.x; // Índice de columna (eje x)
    int j = blockIdx.y * blockDim.y + threadIdx.y; // Índice de fila (eje y)

    // Evitar que hilos fuera del límite de la cuadrícula procesen datos
    if (i >= nx || j >= ny) return;

    int idx = j * nx + i;

    // Condiciones de frontera fijas: mantener los valores de u0 constantes en los bordes
    if (i == 0 || i == nx - 1 || j == 0 || j == ny - 1) {
        u1[idx] = u0[idx];
    }
    // Celdas internas
    else {
        float u_center = u0[idx];
        float u_left   = u0[idx - 1];      // i - 1
        float u_right  = u0[idx + 1];      // i + 1
        float u_up     = u0[idx - nx];     // j - 1 (fila de arriba)
        float u_down   = u0[idx + nx];     // j + 1 (fila de abajo)

        float d2u_dx2 = (u_left - 2.0f * u_center + u_right) / (dx * dx);
        float d2u_dy2 = (u_up - 2.0f * u_center + u_down) / (dy * dy);

        u1[idx] = u_center + kappa * dt * (d2u_dx2 + d2u_dy2);
    }
}
"""

# Compilar el módulo de GPU
mod = SourceModule(heat_eqn_2d_kernel_src)
heatEqn2DGPU = mod.get_function("heatEqn2D")
```

```python
def heatEquation2DGPU(u0, num_timesteps, kappa, dt, dx, dy):
    assert(u0.dtype == np.float32)
    ny, nx = u0.shape

    # Reservar espacio en GPU
    u1_gpu = cuda.mem_alloc(u0.nbytes)
    u0_gpu = cuda.mem_alloc(u0.nbytes)

    # Copiar datos iniciales de CPU a GPU
    cuda.memcpy_htod(u0_gpu, u0)

    # Configuración de bloques e hilos en 2D
    block = (16, 16, 1)
    grid = (int(np.ceil(nx / 16)), int(np.ceil(ny / 16)), 1)

    # Bucle de paso temporal en la GPU
    for n in range(num_timesteps):
        heatEqn2DGPU(
            u1_gpu, u0_gpu,
            np.float32(kappa), np.float32(dx), np.float32(dy), np.float32(dt),
            np.uint32(nx), np.uint32(ny),
            block=block, grid=grid
        )
        # Intercambio de punteros rápidos en GPU (Pointer Swap)
        u0_gpu, u1_gpu = u1_gpu, u0_gpu

    # Copiar el resultado final de vuelta a la CPU
    u1 = np.empty_like(u0)
    cuda.memcpy_dtoh(u1, u0_gpu)

    return u1
```

```python
# Resolver usando la GPU
u_gpu = heatEquation2DGPU(initial_temp.copy(), num_timesteps, kappa, dt, dx, dy)

# Visualizar resultado final obtenido en GPU
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

# Graficar estado inicial
im0 = axes[0].imshow(initial_temp, cmap='jet', origin='lower')
axes[0].set_title("Distribución de Temperatura Inicial")
axes[0].set_xlabel("Malla X")
axes[0].set_ylabel("Malla Y")
fig.colorbar(im0, ax=axes[0], label="Temperatura (°C)")

# Graficar estado final procesado por la GPU
im1 = axes[1].imshow(u_gpu, cmap='jet', origin='lower')
axes[1].set_title(f"Resultado Final en GPU ({num_timesteps} pasos)")
axes[1].set_xlabel("Malla X")
axes[1].set_ylabel("Malla Y")
fig.colorbar(im1, ax=axes[1], label="Temperatura (°C)")

plt.tight_layout()
plt.show()

# Validación rápida de que ambos resultados coinciden
error_diff = np.max(np.abs(u_cpu - u_gpu))
print(f"Diferencia máxima entre CPU y GPU: {error_diff:.6f} °C")
```
