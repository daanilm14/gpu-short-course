```python
import pycuda.gpuarray as gpuarray
import pycuda.autoinit
import numpy as np

# 1. Crear un array en la CPU (NumPy)
A_host = np.array([1, 2, 3, 4], dtype=np.int32)

# 2. Copiar el array a la GPU (Símil a CuArray([1,2,3,4]))
A_device = gpuarray.to_gpu(A_host)

# 3. Operación en GPU (Símil a A_d .+= 1)
A_device += 1

# 4. Traer los datos de vuelta a la CPU (Símil a Array(A_d))
A_host_updated = A_device.get()

print("Resultado en CPU:", A_host_updated)
# Salida esperada: [2, 3, 4, 5]
```
